```python
%%writefile /content/generate_voice.py
import os
os.environ['MPLBACKEND'] = 'Agg'
from TTS.api import TTS
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
tts = TTS('tts_models/multilingual/multi-dataset/xtts_v2').to(device)

# Insert your text here to replace the example "..."
text = """No es cierto. No has dicho nada. Saliste, me encontraste. Saliste y pernoctaste en la mañana a mis pequeñitos quereres que te buscaban cual tortolitos. El amor no es algo si quiera cognoscible, difícil ni paradigmático, me dijiste sin decirlo, como decirlo: con palabras que salen no de una boca. Decías y los tortolitos miraban pasmados tus rasgos, tus pequeñas vidas cenicientas, tus faltas ansiadas de querer, y no se creían lo que te traías encima: un pequeño recuerdo colgado a cada hoyuelo de tu rostro que, al contoneo acompasado de tu sonrisa, probaba la existencia de un amor por decir.
Decías, sin palabras claro está, que dijiste algo. Hablaste, pero no te entendí. Pronunciaste algo como decir que el amor es algo. Algo extravagante. Algo ameno. Y nunca pronunciaste las palabras ameno ni extravagante ni amor. Y mis quereres rebuscaron mi torso, agarraban con sus manitas de dónde colgarse. Escalaron a mi sien, se volcaron, tomaron impulso y apuntaron directo a tus recuerdos: los colgados en tus mejillas y aquellos que yacían rondándote toda. Cayeron. Observaron bien tus verdades, todas faltas sin mentiras, tortolitos quedaron al saber que tus recuerdos eran una extraña forma de querer. Se movían. Se extendían por todo tu rostro, antes de si quiera pedir algún permiso estético. Se bisectaban y volvían a su forma. Ellos sí hablaban con palabras, pero sin voz. Mis quereres no entendían, mas yo sí. Los llamé: regresaron. Los acurruqué donde se aclimatan las pequeñas cosas: en un pequeño cuartito de cincoveinte.
Y no pude escucharte. Sin embargo, veías.
Y a través de esas paredes amargas en las que nos encerramos todos. mis quereres veían que no te escuchaba por primera vez tantas veces. Observaban a tus recuerdos, ahora quereres, que te revoloteaban los cabellos ya aclimatados a ellos. Parecieran aclimatados pues cada curva de fibra capilar era formada por sus manecitas. Cada una de ellas era producto de un querer. Así, cuando te trazaban la frente sin descanso en busca de modelar tus rulos a alguna posición antinatural, esta volvía a rearmarse por aquellos recuerdos, querercitos de pacotilla.
Sin no puedo entender siquiera tu pensamiento. Algo existe que evita que pueda pensar en ti como una persona. Idealizarte es una opción. Cómo duele querer a otros solo por no poder hacerlo hacia tí

A ti:
	Lo confieso. No lo alcanzo a ocultar. A causa de una ligera malformación sentimental me es debido plantear una declaración de delito del cual he de presentarme como único acreedor de culpa.
Es una cosa bien patética, te cuento. Confesar pertenecer a este patetismo clásico que se origina en los desgarros de quien no teme en perderse por su causa delictiva, mas adolece por algo que quiere nacer: ¿de dónde? ¿por qué? ¿por ti?
Justifico: verte me parte entre otras tantas mitades, y, no tenerte, me convierte en tanto hombre mío separado. Luego, cada vez que me levanto, de tanta partición ocasionada, te pienso y mi mundo enflaquece en otras tantas miles dispersiones.
Por eso en esta casi noche oscura, no por ausencia de luna, sino, por la dolencia de un bajo amor al sol que oscurece todo lo demás, me declaro de delito culposo. Culpable de este acto que es la forma más maliciosa de perderse y quizás la única tortuosa forma de vida: quererte amar. Y lo lloraré antes de decírtelo. Y te escribiré tantas cartas en una para que entiendas lo que no podré decirte cuando te busque y no diga cuando lo sientan mis párpados, mis sienes y mis vidas. Y rebuscaré en mis miedos aquellos que te sirvan de excusa. Y me dirás no puedo. Y me dirás quizás. Y me dirás lo siento. Losé.
Y tú quien perteneces al mito, a una realidad extraña. A ese universo quizás igual al mío, que aún de cerca, no puedo ni pensar. Cada vez que intento rozar con la pizca de las casualidades de mis hados, me corto y muero antes de si quiera escribirte o pensarte.
Te escribo este que es un lamento a una enfermedad, condena preventiva de mandato divino reservada para aquellos que no se atreven a confesarse de delito doloso.
Una complacencia brindada por ser victimario confeso de pecado capital.

Sinceramente,
Ernesto."""

# Generate speech
tts.tts_to_file(
    text=text,
    speaker_wav="/content/drive/MyDrive/Audiolectura-Poemas-de-Gemini.wav",  # <-- change this to match your audio file
    language="es",
    file_path="/content/cloned_voice.wav"
)

print("Voice generated: /content/cloned_voice.wav")
```
